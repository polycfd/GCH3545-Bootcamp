## 4.2 Pourquoi la qualité du code est importante

Il est tentant de penser : « Peu importe si mon code est laid, tant qu'il donne le bon résultat. »

Cette idée est dangereuse.

Un code difficile à lire peut fonctionner aujourd'hui, mais devenir très problématique lorsque :

- vous devez le modifier ;
- vous devez expliquer votre démarche ;
- vous devez trouver une erreur ;
- vous devez comparer plusieurs cas ;
- un collègue ou un chargé de TD doit comprendre ce que vous avez fait ;
- vous revenez au code plusieurs semaines plus tard.

Dans un cours de modélisation numérique, la qualité du code fait partie de la démarche scientifique.

Un résultat numérique n'est crédible que si le code qui l'a produit peut être compris et vérifié.

## 4.4 Améliorer les noms de variables

La première amélioration consiste à utiliser des noms de variables explicites. Un bon nom de variable indique ce que la variable représente.

Comparons :

```python
a = 500
```

avec :

```python
temperature_initiale = 500
```

La deuxième version est plus longue, mais elle est beaucoup plus claire.

### ▶ À faire

Exécutez la cellule suivante.

````python
temperature_initiale = 500.0       # [°C]
temperature_ambiante = 25.0        # [°C]
coefficient_refroidissement = 0.03 # [1/s]

temps = np.linspace(0, 200, 201)   # [s]

temperatures = (
    temperature_ambiante
    + (temperature_initiale - temperature_ambiante)
    * np.exp(-coefficient_refroidissement * temps)
)

plt.plot(temps, temperatures)
plt.xlabel("Temps [s]")
plt.ylabel("Température [°C]")
plt.title("Refroidissement d'une sphère chaude")
plt.grid(True)
plt.show()
````

### ⚠ Réflexion

Comparez cette version avec la précédente.

- Quelle version est plus longue ?
- Quelle version est plus facile à comprendre ?
- Où les unités sont-elles indiquées ?
- Si vous revenez à ce code dans deux mois, laquelle des deux versions préférerez-vous relire ?

En programmation scientifique, un code légèrement plus long mais beaucoup plus clair est souvent préférable à un code compact mais opaque.

## 4.5 Écrire une fonction pour encapsuler le modèle

Dans la Partie 3, nous avons déjà introduit une fonction pour le modèle de température.

Reprenons cette idée.

Une fonction permet de donner un nom à un calcul et de le réutiliser.

### ▶ À faire

Étudiez la cellule suivante.

````python
def temperature_sphere(t, temperature_initiale, temperature_ambiante, coefficient_refroidissement):
    """
    Calcule la température d'une sphère chaude en refroidissement.

    Paramètres
    ----------
    t : float ou ndarray
        Temps [s].
    temperature_initiale : float
        Température initiale de la sphère [°C].
    temperature_ambiante : float
        Température ambiante [°C].
    coefficient_refroidissement : float
        Coefficient de refroidissement [1/s].

    Retour
    ------
    float ou ndarray
        Température de la sphère [°C].
    """
    return (
        temperature_ambiante
        + (temperature_initiale - temperature_ambiante)
        * np.exp(-coefficient_refroidissement * t)
    )
````

### ⚠ Réflexion

Cette fonction est plus longue que la formule mathématique.

Mais elle est aussi plus utile :

- elle donne un nom clair au calcul ;
- elle indique les unités ;
- elle peut être appelée plusieurs fois ;
- elle peut être testée ;
- elle sépare le modèle du reste du programme.

Une bonne fonction est souvent la première étape vers un code plus fiable.

## 4.6 Une fonction doit avoir une responsabilité claire

Une fonction devrait idéalement faire une seule chose.

Par exemple, la fonction `temperature_sphere` calcule une température.

Elle ne trace pas de graphique.  
Elle ne choisit pas les paramètres.  
Elle ne sauvegarde pas de fichier.  
Elle ne demande pas d'entrée à l'utilisateur.

C'est volontaire.

Si une fonction fait trop de choses, elle devient difficile à comprendre et difficile à tester.

### Exemple moins souhaitable

La fonction suivante calcule la température et trace aussi le graphique.

Elle fonctionne, mais elle mélange deux responsabilités différentes.

````python
def calculer_et_tracer_temperature(temps, T0, T_inf, k):
    temperatures = T_inf + (T0 - T_inf) * np.exp(-k * temps)

    plt.plot(temps, temperatures)
    plt.xlabel("Temps [s]")
    plt.ylabel("Température [°C]")
    plt.title("Refroidissement")
    plt.show()

    return temperatures

temperatures_test = calculer_et_tracer_temperature(temps, 500, 25, 0.03)
````

### ⚠ Réflexion

Cette fonction est-elle pratique si vous voulez seulement calculer les températures sans produire de graphique ?

Est-elle pratique si vous voulez tester uniquement la formule ?

Pas vraiment.

Il est souvent préférable de séparer :

- le calcul ;
- la visualisation ;
- la vérification.

### Version mieux structurée

On peut écrire une fonction pour le calcul et une autre pour le graphique.

````python
def tracer_temperature(temps, temperatures, temperature_ambiante=None):
    """
    Trace l'évolution de la température en fonction du temps.

    Paramètres
    ----------
    temps : ndarray
        Temps [s].
    temperatures : ndarray
        Températures [°C].
    temperature_ambiante : float, optional
        Température ambiante [°C]. Si fournie, une ligne horizontale est tracée.
    """
    plt.plot(temps, temperatures, label="Température de la sphère")

    if temperature_ambiante is not None:
        plt.axhline(temperature_ambiante, linestyle="--", label="Température ambiante")

    plt.xlabel("Temps [s]")
    plt.ylabel("Température [°C]")
    plt.title("Refroidissement d'une sphère chaude")
    plt.grid(True)
    plt.legend()
    plt.show()


temps = np.linspace(0, 200, 201)
temperatures = temperature_sphere(temps, 500, 25, 0.03)

tracer_temperature(temps, temperatures, temperature_ambiante=25)
````

### 💡 À retenir

Une fonction courte avec une responsabilité claire est généralement :

- plus facile à lire ;
- plus facile à corriger ;
- plus facile à réutiliser ;
- plus facile à tester.

Cette idée est centrale en programmation scientifique.

## 4.7 Commentaires et docstrings

Les commentaires et les docstrings sont utiles, mais ils ne servent pas au même objectif.

### Docstring

Une docstring décrit une fonction :

- ce qu'elle fait ;
- ses paramètres ;
- ce qu'elle retourne ;
- les unités importantes.

Elle est placée juste après la ligne `def`.

### Commentaire

Un commentaire explique une ligne ou une décision particulière dans le code.

Il devrait expliquer **pourquoi** quelque chose est fait, surtout si ce n'est pas évident.

### Commentaire peu utile

```python
x = x + 1  # ajoute 1 à x
```

Ce commentaire répète simplement le code.

### Commentaire plus utile

```python
nombre_iterations += 1  # on compte cette itération avant de tester l'arrêt
```

Ici, le commentaire explique une intention.

### 💡 À retenir

Un commentaire ne devrait pas compenser un mauvais nom de variable.

Comparez :

```python
a = 500  # température initiale
```

avec :

```python
temperature_initiale = 500
```

La deuxième version est préférable. Le meilleur commentaire est parfois un meilleur nom de variable.

## 4.8 Éviter la répétition inutile

La répétition de code est une source fréquente d'erreurs.

Supposons que vous écriviez la formule de refroidissement à plusieurs endroits :

```python
T1 = T_inf + (T0 - T_inf) * np.exp(-k1 * temps)
T2 = T_inf + (T0 - T_inf) * np.exp(-k2 * temps)
T3 = T_inf + (T0 - T_inf) * np.exp(-k3 * temps)
```

Cela fonctionne, mais si vous devez modifier la formule, vous devez penser à la modifier partout.

Il est préférable d'utiliser une fonction.

### ▶ À faire

Exécutez la cellule suivante.

````python
valeurs_k = [0.01, 0.03, 0.08]

for coefficient in valeurs_k:
    temperatures = temperature_sphere(temps, temperature_initiale=500, temperature_ambiante=25, coefficient_refroidissement=coefficient)
    plt.plot(temps, temperatures, label=f"k = {coefficient} 1/s")

plt.axhline(25, linestyle="--", label="Température ambiante")
plt.xlabel("Temps [s]")
plt.ylabel("Température [°C]")
plt.title("Influence du coefficient de refroidissement")
plt.grid(True)
plt.legend()
plt.show()
````

### ⚠ Réflexion

La formule du modèle apparaît maintenant à un seul endroit : dans la fonction `temperature_sphere`.

C'est une bonne chose.

Si nous corrigeons ou modifions le modèle, nous n'avons qu'un seul endroit à modifier.

Cette idée réduit le risque d'incohérence entre différentes parties du programme.

## 4.10 Structure de projet recommandée

Jusqu'ici, nous avons surtout travaillé dans un notebook.

Pour un petit exercice, c'est très bien. Pour un projet plus long, il est utile d'organiser les fichiers.

Une structure simple pourrait être :

```text
projet_modele/
│
├── src/
│   └── modele.py
│
├── notebooks/
│   └── exploration.ipynb
│
├── tests/
│   └── test_modele.py
│
├── resultats/
│
└── README.md
```

Chaque dossier a un rôle :

| Élément | Rôle |
|---|---|
| `src/` | code source réutilisable |
| `notebooks/` | exploration, visualisation, analyse |
| `tests/` | tests automatisés |
| `resultats/` | figures, tableaux, sorties |
| `README.md` | description du projet |

### ⚠ Réflexion

Pourquoi ne pas tout mettre dans un seul fichier ?

Parce qu'un projet grandit rapidement.

Une bonne structure permet de retrouver plus facilement :

- le modèle ;
- les tests ;
- les figures ;
- les résultats ;
- les explications.

Elle facilite aussi le travail d'équipe.
