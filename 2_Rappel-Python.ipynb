## 2.6 Documenter une fonction avec une _docstring_

Une fonction peut contenir une courte documentation appelée **docstring**.

Cette documentation explique :

- ce que fait la fonction ;
- quels sont ses paramètres ;
- ce qu'elle retourne ;
- quelles unités sont utilisées.

Cela peut sembler un peu formel pour une petite fonction, mais c'est une excellente habitude en programmation scientifique.

### ▶ À faire

Exécutez la cellule suivante.

````python
def masse_sphere(rayon, masse_volumique):
    """
    Calcule la masse d'une sphère.

    Paramètres
    ----------
    rayon : float
        Rayon de la sphère [m].
    masse_volumique : float
        Masse volumique du matériau [kg/m³].

    Retour
    ------
    float
        Masse de la sphère [kg].
    """
    volume = 4/3 * np.pi * rayon**3
    return masse_volumique * volume

masse = masse_sphere(0.01, 7800)

print(masse)
````

### ⚠ Réflexion

La fonction précédente est plus longue que strictement nécessaire.

Mais elle est aussi plus claire :

- on sait quelles unités utiliser ;
- on sait ce que représente chaque paramètre ;
- on sait ce que la fonction retourne.

Dans le cadre d'un projet de modélisation numérique, cette clarté revêt une importance capitale, par exemple pour détecter des erreurs, si l'on souhaite modifier le code ultérieurement, ou si ce code doit être compréhensible pour d'autres utilisateurs.
