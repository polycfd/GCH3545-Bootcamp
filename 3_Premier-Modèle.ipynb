## 3.3 Le modèle mathématique

Pour décrire le refroidissement de la sphère, nous utiliserons un modèle simple :

$T(t) = T_\infty + (T_0 - T_\infty)\exp(-kt)$

où :

| Symbole | Signification |
|---|---|
| $T(t)$ | température de la sphère au temps $t$ |
| $T_0$ | température initiale |
| $T_\infty$ | température ambiante |
| $k$ | coefficient de refroidissement |
| $t$ | temps |

Le paramètre $k$ contrôle la vitesse de refroidissement. Un grand $k$ correspond à un refroidissement rapide, un petit $k$ correspond à un refroidissement lent.

Dans cette partie, nous ne dérivons pas cette équation. Nous l'utilisons comme modèle simple afin de nous concentrer sur la traduction du modèle en code Python.

### ⚠ Réflexion

Avant de coder, vérifions si le modèle semble cohérent.

Si `t = 0`, alors :

$\exp(-k \cdot 0) = \exp(0) = 1$

Donc :

$T(0) = T_\infty + (T_0 - T_\infty) = T_0$

C'est exactement ce que l'on attend : au début, la température doit être égale à la température initiale.

Par contre, si `t = 10` et `k = 1`:

$\exp(-10) = 4.54 \times 10^{-5} \approx 0$

Donc :

$T(10) \approx T_\infty$

C'est exactement ce que l'on attend : au bout d'un certain temps, la température doit se rapprocher de la température ambiante.

Cette observation nous donnera une première vérification simple du code.

## 3.8 Distinguer modèle, code et résultats

À ce stade, nous avons plusieurs objets différents :

### (i) Le problème physique

Une sphère chaude refroidit dans un environnement plus froid.

### (ii) Le modèle mathématique

$T(t) = T_\infty + (T_0 - T_\infty)\exp(-kt)$

### (iii) L'implémentation Python

```python
def temperature_sphere(t, T0, T_inf, k):
    return T_inf + (T0 - T_inf) * np.exp(-k * t)
```

### (iv) Les résultats

Un tableau de températures et un graphique.

**Ces quatre niveaux sont liés, mais ils ne sont pas identiques.**

Une erreur peut apparaître à chacun de ces niveaux :

- le modèle physique peut être trop simplifié ;
- l'équation peut être mal comprise ;
- le code peut contenir une erreur ;
- le graphique peut être mal interprété.

C'est pourquoi la vérification et l'interprétation sont essentielles.
