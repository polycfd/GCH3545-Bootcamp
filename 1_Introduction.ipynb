### 💡 À retenir : `assert`

Vous venez de rencontrer la commande `assert`. Elle permet de vérifier automatiquement qu'une condition est satisfaite.

Exemple :

```python
assert temperature > 0
```

Cela signifie : « Je m'attends à ce que la température soit positive. Si ce n'est pas le cas, il y a probablement un problème. »

Si la condition est vraie, l'exécution continue normalement. Si la condition est fausse, Python arrête l'exécution et affiche une erreur.

Nous utiliserons régulièrement `assert` dans ce bootcamp pour vérifier automatiquement les exercices. Plus tard, cette idée sera reliée aux **tests automatisés**, qui sont essentiels pour vérifier le comportement d'un programme.

## 1.4 Variables et informations

## 1.7 Premier calcul scientifique avec NumPy

En modélisation numérique, nous devons souvent effectuer un même calcul pour plusieurs valeurs d'une variable. Par exemple :

- calculer une température à plusieurs instants ;
- calculer une concentration à plusieurs positions ;
- calculer une pression pour différentes conditions opératoires ;
- comparer plusieurs valeurs d'un paramètre.

Pour cela, NumPy permet de créer facilement des tableaux de valeurs.

L'une des fonctions les plus utiles est :

```python
np.linspace(debut, fin, nombre_de_points)
```

Cette fonction crée des valeurs **également espacées** entre deux bornes.

Par exemple :

```python
np.linspace(0, 10, 11)
```

demande à NumPy de créer 11 valeurs entre 0 et 10, en incluant les deux bornes.

### ▶ À faire

Exécutez la cellule suivante.

````python
x = np.linspace(0, 10, 11)

print(x)
````

Ce type de vérification simple est important : avant de faire confiance à un calcul plus complexe, on vérifie d'abord que les données de base sont bien celles que l'on croit.

Créez maintenant un tableau nommé `temps` contenant 6 valeurs entre 0 et 5 inclusivement.

### ⚠ Vérification automatique

La cellule suivante vérifie que :

- `temps` est bien un tableau NumPy ;
- il contient 6 valeurs ;
- il commence à 0 ;

````python
assert isinstance(temps, np.ndarray), "temps doit être un tableau NumPy."
assert len(temps) == 6, "temps doit contenir 6 valeurs."
assert np.isclose(temps[0], 0), "La première valeur doit être 0."
assert np.isclose(temps[-1], 5), "La dernière valeur doit être 5."

print("✅ Exercice 1.3 réussi")
````

## 1.8 Premier graphique

Une grande partie de la modélisation numérique consiste à analyser des résultats sous forme graphique.

Un tableau de nombres est utile, mais un graphique permet souvent de voir immédiatement :

- une tendance ;

Nous allons produire notre premier graphique avec Matplotlib.

## 1.9 Scripts Python

## 1.10 Demander de l'aide efficacement

Pendant le trimestre, vous rencontrerez des erreurs. C'est normal. Même les programmeurs expérimentés passent beaucoup de temps à lire des messages d'erreur et à corriger des problèmes.

Un message d'erreur ne signifie pas que vous êtes mauvais en programmation. Il signifie que Python a rencontré quelque chose qu'il ne comprend pas ou qu'il ne peut pas exécuter.

La qualité de votre question influence fortement la qualité de l'aide que vous recevrez.

### ❌ Peu utile

```text
Mon code ne marche pas.
```

Cette phrase ne donne pas assez d'information pour comprendre le problème.

### ✅ Beaucoup plus utile

Lorsque vous demandez de l'aide, essayez de fournir :

- le message d'erreur complet ;
- le code concerné ;
- le résultat attendu ;
- le résultat obtenu ;
- ce que vous avez déjà essayé.

Par exemple :

```text
J'obtiens l'erreur suivante :
NameError: name 'temperature_initial' is not defined

Voici la ligne concernée :
print(temperature_initial)

Je pensais afficher la température initiale, qui devrait valoir 500.
```

Cette façon de demander de l'aide fonctionne avec :

- les chargés de TD ;
- le professeur ;
- vos collègues ;
- les forums ;
- les outils d'IA générative.

### ✍ Exercice 1.5

La cellule suivante contient volontairement une erreur.

Exécutez-la, observez le message d'erreur, puis essayez d'identifier le problème avant de regarder la correction.

````python
temperature_initiale = 500

print(temperature_initial)
````

### ⚠ Réflexion

Python devrait afficher une erreur du type :

```text
NameError: name 'temperature_initial' is not defined
```

Cette erreur signifie que Python ne connaît pas le nom `temperature_initial`.

Comparez attentivement :

```python
temperature_initiale
temperature_initial
```

Les deux noms se ressemblent, mais ils ne sont pas identiques.

En programmation, une seule lettre manquante suffit à créer une erreur.

### ▶ Correction

Exécutez maintenant la cellule corrigée.

````python
temperature_initiale = 500

print(temperature_initiale)
````

### 💡 À retenir

Les erreurs de noms de variables sont très fréquentes.

Pour les éviter :

- utilisez des noms clairs ;
- soyez cohérent ;
- évitez de changer inutilement de notation ;
- lisez attentivement les messages d'erreur.

Un message d'erreur est souvent une source d'information utile, pas seulement un obstacle.
