## 6.2 Usages appropriés de l'IA

Un outil d'IA peut être très utile lorsqu'il est utilisé comme **assistant**. Voici quelques usages raisonnables dans un cours de modélisation numérique.

### Expliquer une erreur

Par exemple, vous pouvez demander :

```text
J'obtiens l'erreur suivante en Python :
NameError: name 'temperature_initial' is not defined

Peux-tu m'expliquer ce que cela signifie ?
```

### Proposer une structure de fonction

Par exemple :

```text
Je veux écrire une fonction Python qui calcule la température d'une sphère en refroidissement.
Quels paramètres cette fonction devrait-elle recevoir ?
```

### Améliorer la lisibilité

Par exemple :

```text
Peux-tu proposer de meilleurs noms de variables pour ce code ?
```

### Proposer des tests

Par exemple :

```text
Quels tests simples pourrais-je écrire pour vérifier un modèle de refroidissement exponentiel ?
```

### Expliquer une bibliothèque

Par exemple :

```text
Que fait np.linspace(0, 10, 101) ?
```

Dans tous ces cas, l'IA aide à comprendre, structurer ou vérifier. Elle ne remplace pas votre jugement.

### 💡 À retenir

Un bon usage de l'IA est souvent un usage où vous restez actif :

- vous posez une question précise ;
- vous lisez la réponse ;
- vous vérifiez le résultat ;
- vous adaptez la solution ;
- vous êtes capable de l'expliquer.

##### **L'IA est un outil d'aide, pas une autorité scientifique.**

## 6.3 Usages risqués ou inappropriés

Certains usages de l'IA sont beaucoup plus problématiques.

### Exemple 1 : soumettre du code non compris

Demander à un outil d'IA de produire une solution complète, puis la soumettre sans la comprendre, est une mauvaise pratique.

Même si le code fonctionne, vous ne pouvez pas garantir qu'il est correct.

### Exemple 2 : faire confiance aux résultats sans vérification

Un outil d'IA peut produire une fonction qui semble plausible, mais qui contient une erreur de signe, d'unité ou de logique.

### Exemple 3 : inventer des hypothèses physiques

L'IA peut proposer une équation ou une hypothèse qui semble raisonnable, mais qui n'est pas adaptée au problème.

### Exemple 4 : inventer des références

Les outils d'IA peuvent parfois inventer des références bibliographiques, des équations, des noms d'auteurs ou des sources.

### Exemple 5 : masquer l'utilisation de l'IA

Si un cours, un laboratoire, une entreprise ou un journal demande de déclarer l'utilisation d'outils d'IA, il faut respecter cette règle.

##### **L'intégrité professionnelle ne se limite pas au résultat final. Elle concerne aussi la manière dont le travail a été produit.**

## 6.4 Un workflow responsable

Un usage responsable de l'IA suit un processus. Voici un workflow recommandé :

- Demander de l'aide à l'IA
- Lire le code ou l'explication
- Expliquer le code avec vos propres mots
- Vérifier le comportement
- Tester
- Documenter

L'étape importante est que l'IA n'est pas la dernière étape. L'IA peut proposer une solution, mais vous devez ensuite vérifier cette solution.

### 💡 À retenir

## 6.5 Exemple : code généré par IA

Imaginons qu'un outil d'IA nous propose la fonction suivante pour le refroidissement d'une sphère.

Le code est bien présenté. Il contient une docstring. Il semble raisonnable.

### ▶ À faire

Exécutez la cellule suivante.

````python
import numpy as np
import matplotlib.pyplot as plt

def temperature_sphere_ia(t, temperature_initiale, temperature_ambiante, coefficient_refroidissement):
    """
    Calcule la température d'une sphère en refroidissement selon un modèle exponentiel.

    Paramètres
    ----------
    t : float ou ndarray
        Temps [s].
    temperature_initiale : float
        Température initiale [°C].
    temperature_ambiante : float
        Température ambiante [°C].
    coefficient_refroidissement : float
        Coefficient de refroidissement [1/s].

    Retour
    ------
    float ou ndarray
        Température [°C].
    """
    return temperature_ambiante + temperature_initiale * np.exp(-coefficient_refroidissement * t)
````

### ⚠ Réflexion


Mais le modèle attendu est : $T(t) = T_\infty + (T_0 - T_\infty)\exp(-kt)$

Voyez-vous une différence ?

## 6.8 Comment formuler une bonne demande à l'IA

La qualité de la réponse dépend souvent de la qualité de la question. Une demande vague produit souvent une réponse vague.

### Demande faible

```text
Fais mon code.
```

Cette demande ne précise ni le modèle, ni les unités, ni les attentes, ni les vérifications.

### Demande plus utile

```text
Je veux écrire une fonction Python pour le modèle T(t) = T_inf + (T0 - T_inf) exp(-k t).

Les unités sont :
- t en secondes ;
- T0 et T_inf en degrés Celsius ;
- k en 1/s.

Peux-tu proposer une fonction claire avec une docstring, puis suggérer trois tests simples pour vérifier son comportement ?
```

Cette deuxième demande est beaucoup plus précise.

Elle indique :

- le modèle ;
- les variables ;
- les unités ;
- le format attendu ;
- le besoin de vérification.

### ⚠ Réflexion

Une bonne demande ne sert pas seulement à obtenir une meilleure réponse de l'IA. Elle vous force aussi à clarifier votre propre compréhension du problème.

Si vous n'arrivez pas à expliquer clairement ce que vous voulez demander, c'est souvent que le problème n'est pas encore assez bien compris.

## 6.9 Intégrité académique, confidentialité et propriété intellectuelle

L'utilisation responsable de l'IA ne concerne pas seulement la qualité du code. Elle concerne aussi l'intégrité académique, la confidentialité et la propriété intellectuelle.

### Intégrité académique

Respectez toujours les règles du cours, du département et de l'université.

L'utilisation d'un outil d'IA ne remplace pas votre responsabilité de :

- comprendre le code ;
- vérifier les résultats ;
- produire un travail que vous êtes capable d'expliquer ;
- déclarer l'utilisation de l'IA si cela est demandé.

### Confidentialité

Ne transmettez pas à un outil d'IA :

- des données confidentielles ;
- des données personnelles ;
- des informations provenant d'un partenaire industriel ;
- des informations soumises à une entente de confidentialité ;
- des résultats de recherche non publiés si leur diffusion n'est pas autorisée.

Une fois une information partagée avec un outil externe, vous ne contrôlez pas toujours complètement son utilisation, sa conservation ou sa diffusion.

### Propriété intellectuelle

Ne supposez pas qu'il est acceptable de transmettre n'importe quel code à un outil d'IA.

Il peut être inapproprié de partager :

- un code propriétaire d'entreprise ;
- un logiciel commercial ;
- un dépôt Git privé ;
- un projet de recherche non publié ;
- du matériel soumis à des restrictions de diffusion ;
- du code appartenant à un client, un laboratoire ou un partenaire industriel.

**Avant d'utiliser un outil d'IA avec un projet réel, vérifiez toujours les règles applicables à votre organisation.**

### 💡 À retenir

L'utilisation responsable de l'IA ne consiste pas seulement à obtenir du code fonctionnel.

Elle consiste aussi à respecter les principes de :

- vérification ;
- confidentialité ;
- propriété intellectuelle ;
- intégrité académique ;
- intégrité professionnelle.

## 6.10 Limites de l'IA

Les outils d'IA sont puissants, mais ils ont des limites.

Ils peuvent :

- produire une réponse plausible mais fausse ;
- oublier une hypothèse importante ;
- utiliser une convention différente de celle du cours ;
- mélanger des unités ;
- proposer un code trop complexe ;
- inventer une source ;
- ignorer des contraintes spécifiques ;
- donner une réponse différente à la même question.

C'est pourquoi il faut rester critique.

Un outil d'IA peut accélérer certaines tâches, mais il ne remplace pas :

- la compréhension du problème ;
- le raisonnement physique ;
- la vérification ;
- la validation ;
- le jugement professionnel.

### ⚠ Réflexion

Réponse courte : enquêter.

Il ne faut pas supposer automatiquement que l'IA a raison.  
Il ne faut pas non plus supposer automatiquement que vos tests sont parfaits.

Vous devez comprendre la cause du désaccord.
